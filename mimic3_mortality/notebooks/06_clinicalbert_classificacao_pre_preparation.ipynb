```python
import pandas as pd
```

```python
# 1. Carrega NOTEEVENTS (sumários clínicos)
df_notes = pd.read_csv("../data/Mimic-3_patients/NOTEEVENTS/NOTEEVENTS_sorted.csv")
```

```python
# 2. Carrega ADMISSIONS para pegar a variável alvo
df_adm = pd.read_csv("../data/Mimic-3_patients/ADMISSIONS/ADMISSIONS_sorted.csv")
```

```python
# 3. Junta as duas tabelas pela HADM_ID
df = df_notes.merge(df_adm[['HADM_ID', 'HOSPITAL_EXPIRE_FLAG']], on='HADM_ID', how='inner')
```

```python
# 4. Renomeia colunas para consistência
df = df.rename(columns={"TEXT": "text", "HOSPITAL_EXPIRE_FLAG": "hospital_expire_flag"})
```

```python
# 5. Filtra apenas os registros válidos (ex: notas do tipo 'Discharge summary')
df = df[df['CATEGORY'] == 'Discharge summary']
```

```python
# 6. Remove nulos
df = df[['text', 'hospital_expire_flag']].dropna()
```

```python
# salvar
df.to_csv("../models/discharge_summaries_labeled.csv", index=False)
```
