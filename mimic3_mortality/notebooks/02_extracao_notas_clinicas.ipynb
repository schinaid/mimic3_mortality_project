# 📄 Notebook 02 - Extração de Notas Clínicas

Este notebook realiza a extração de sumários de alta do dataset `NOTEEVENTS`, utilizando apenas as notas da categoria `"Discharge summary"`. 
O objetivo é preparar os dados textuais clínicos para análises posteriores (NLP, classificação, clusterização etc.).

As etapas incluem:
- Carregamento dos dados
- Filtragem das notas relevantes
- Contagem de palavras por nota
- Salvamento do conjunto processado


```python
import pandas as pd
import os

def load_discharge_notes(file_path):
    """Carrega o arquivo NOTEEVENTS e filtra apenas notas de 'Discharge summary'."""
    df = pd.read_csv(file_path)
    df = df[df['CATEGORY'] == 'Discharge summary'].copy()
    df.reset_index(drop=True, inplace=True)
    return df

def compute_word_counts(df):
    """Adiciona coluna com contagem de palavras por nota."""
    df['NUM_WORDS'] = df['TEXT'].fillna("").apply(lambda x: len(x.split()))
    return df

# Caminho
file_path = "../data/Mimic-3_patients/NOTEEVENTS/NOTEEVENTS_random.csv"
df_notes = load_discharge_notes(file_path)
df_notes = compute_word_counts(df_notes)

df_notes.head(2)
```

```python
print("Total de sumários de alta extraídos:", len(df_notes))
print("\nResumo estatístico da contagem de palavras:")
print(df_notes['NUM_WORDS'].describe())
```

```python
output_path = "../models/discharge_summaries_labeled.csv"
os.makedirs(os.path.dirname(output_path), exist_ok=True)
df_notes.to_csv(output_path, index=False)
print(f"✅ Arquivo salvo em: {output_path}")
```

### ✅ Conclusão

- Foram extraídos **13063** sumários de alta da categoria `"Discharge summary"`.
- A contagem média de palavras por nota foi de aproximadamente **1200 palavras**.
- O conjunto foi salvo em `../models/discharge_summaries_labeled.csv` para uso nos próximos notebooks.
