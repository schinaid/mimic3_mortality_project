# 🔬 Notebook 03 - Extração de Entidades Clínicas com scispaCy

Este notebook aplica técnicas de Processamento de Linguagem Natural com o modelo `en_core_sci_sm` da biblioteca `scispaCy` para extrair entidades clínicas de sumários de alta da base MIMIC-III.

**Objetivo**: Identificar automaticamente entidades médicas relevantes em textos clínicos desestruturados como parte da análise de mortalidade hospitalar.

---



```python
import pandas as pd
import spacy
from collections import Counter

def carregar_texto_sumario(path: str, index: int = 0) -> str:
    """Carrega o texto de um sumário de alta pelo índice da linha."""
    df = pd.read_csv(path)
    return df.loc[index, "TEXT"]
```

```python
def carregar_modelo_scispacy(model_name: str = "en_core_sci_sm"):
    """Carrega o modelo do scispaCy."""
    try:
        return spacy.load(model_name)
    except OSError as e:
        print(f"Erro ao carregar modelo: {e}")
        return None
```

```python
def extrair_entidades(texto: str, nlp_model) -> list:
    """Extrai entidades nomeadas de um texto clínico."""
    doc = nlp_model(texto)
    return [(ent.text, ent.label_) for ent in doc.ents]
```

```python
# Caminho para o arquivo de sumários
path_sumarios = "../data/Mimic-3_patients/NOTEEVENTS/NOTEEVENTS_random.csv"

# Carregamento do texto
texto = carregar_texto_sumario(path_sumarios, index=0)

# Carrega o modelo clínico
nlp = carregar_modelo_scispacy()

# Extrai entidades
entidades = extrair_entidades(texto, nlp)

# Contagem e amostra
contador = Counter(label for _, label in entidades)
print("Contagem de entidades:", contador)
print("Amostra:", entidades[:20])
```
